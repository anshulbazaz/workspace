```python
import numpy as np
import time
```

```python
## Binary search can be done in to ways -
## 1. Using a loop 
## 2. Using a recursive function
## Pre-requisite - the array or list should be sorted and distinct
```

```python
#Binary search using a loop 
def binary_search_loop(num_list, num_to_find):

    l = 0
    r = len(num_list) - 1

    while l <= r:
        m = (l + r) // 2
        m_num = num_list[m]

        if(m_num == num_to_find):
            return m

        if(m_num < num_to_find):
            l = m+1

        if(m_num > num_to_find):
            r = m-1

    return -1
```

```python
#Binary search using a recursive function
def binary_search_recursive(num_list, num_to_find, l, r):

    if(l > r):
        return -1

    m = (l + r) // 2

    if((m > len(num_list)) or (m < 0)):
        return -1

    m_num = num_list[m]

    if(m_num == num_to_find):
        return m

    if(m_num < num_to_find):
        l = m + 1

    if(m_num > num_to_find):
        r = m-1

    return binary_search_recursive(num_list, num_to_find, l, r)
```

```python
temp = np.random.randint(300000, size=(300000))
num_list = list(set(temp))
num_list.sort()
num_to_find = num_list[-1]

start_time = time.time()
ind = binary_search_loop(num_list, num_to_find)
end_time = time.time()
print(f"Number found at {ind} index in {end_time-start_time:.20f} seconds using loop method")

start_time = time.time()
ind = binary_search_recursive(num_list, num_to_find, 0, len(num_list)-1)
end_time = time.time()

print(f"Number found at {ind} index in {end_time - start_time:.20f} seconds using recursive method")
```
