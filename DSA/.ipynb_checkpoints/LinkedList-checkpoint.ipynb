```python
import numpy as np
```

```python
## Creating a linked list object

## Defining each node on the linked list
class linkedlist_node:
    def __init__(self, data=None, next=None):
        self.data = data
        self.next = next
```

```python
## Defining the linked list and  various operations on the linked list
class linkedlist:
    def __init__(self):
        self.head = None  # initializing the linked list with an empty head

    # Inserting data at the beginning of the list
    def insert_beginning(self, data):
        node = linkedlist_node(data, self.head)
        self.head = node

    def insert_end(self, data):
        if self.head is None:
            self.head = linkedlist_node(data, None)  # If linked list is empty, then insert the data with no next
            return

        itr = self.head
        while (itr.next):
            itr = itr.next  # iterate till we reach to the last element

        itr.next = linkedlist_node(data, None)  # At the last element, point it to the new element to be inserted

    def print_linked_list(self):
        if self.head is None:
            print("Linked list is empty")  # If linked list is empty, then insert the data with no next
            return
        print_list = ''
        itr = self.head
        while(itr):
            print_list += str(itr.data) + '-->' if itr.next else str(itr.data)
            itr = itr.next

        print(print_list)

    def get_len(self):
        if self.head is None:
            return 0

        count = 0
        itr = self.head
        while(itr):
            count += 1
            itr = itr.next

        return count

    def insert_at(self, data, index):
        if((index < -1) | (index > self.get_len())):
            print("invalid index")
            return

        if index == -1:
            self.insert_end(data)
            return

        if index == 0:
            self.insert_beginning(data)
            return

        count = 0
        itr = self.head
        while(itr):
            if(count == index - 1):
                node = linkedlist_node(data, itr.next)
                itr.next = node

            itr =  itr.next
            count += 1


    def remove_at(self, index):

        if ((index < 0) | (index >= self.get_len())):
            print("invalid index")
            return

        if index == 0:
            self.head = self.head.next
            return

        itr = self.head
        count = 0

        while(itr):

            if(count == index-1):
                itr.next = itr.next.next
                break

            count = count+1
            itr = itr.next
```

```python
ll = linkedlist()
ll.insert_beginning('a')
ll.insert_beginning(0)
ll.insert_end('b')
ll.insert_at('c', -1)
ll.insert_at('d',2)
ll.print_linked_list()
print(ll.get_len())
ll.remove_at(0)
ll.remove_at(1)
ll.print_linked_list()
print(ll.get_len())
```
