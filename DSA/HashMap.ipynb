## A HashMap is a data structure that is used to store and retrieve values based on keys.
### Fast access time: HashMaps provide constant time access to elements, which means that retrieval and insertion of elements are very fast, usually O(1) time complexity.
### Below hashmap also takes care of collision cases (Different keys pointing to the same index)

```python
class Hashmap:

    def __init__(self):
        self.MAX = 1000
        self.arr = [[] for i in range(self.MAX)]

    def get_hash(self, key):
        hash = 0
        for char in key:
            hash += ord(char)

        return hash % self.MAX

    def __getitem__(self, key):
        arr_index = self.get_hash(key)
        for kv in self.arr[arr_index]:
            if kv[0] == key:
                return kv[1]

    def __setitem__(self, key, value):
        arr_index = self.get_hash(key)
        found = False
        for i,j in enumerate(self.arr[arr_index]):
            if((len(j) == 2) and (j[0] == key)):
                found = True
                self.arr[arr_index][i] = (key,value)

        if(not found):
            self.arr[arr_index].append((key,value))

    def __delitem__(self, key):
        arr_index = self.get_hash(key)
        for i,j in enumerate(self.arr[arr_index]):
            if j[0] == key:
                del self.arr[arr_index][i]
                print(key, " deleted")
```

```python
sales = Hashmap()
sales['march 1'] = 999
sales['march 2'] = 1999
sales['march 6'] = 2100
sales['march 17'] = 200

print(sales['march 2'])

del sales['march 2']

print(sales['march 17'])
```
